Prompt Oluşturma Fonksiyonu

```python
def make_prompt(soru: str, dogru_cevap: str, ogrenci_cevap: str, sbert_qs: float, sbert_sc: float):
    # .4f kullanarak sayıların modelin alışık olduğu (0.1234 gibi) formatta kalmasını sağlarız
    prompt = f"""### Instruction:
Öğrenci cevabını analiz et, SBERT benzerlik skorlarını değerlendir ve 0.0-1.0 arasında bir puan ver.

### Input:
Soru: {soru}
Doğru Cevap: {dogru_cevap}
Öğrenci Cevabı: {ogrenci_cevap}
SBERT Skorları: Soru-Öğrenci Cevabı Benzerliği: {sbert_qs:.4f}, Öğrenci Cevabı-Doğru Cevap benzerliği: {sbert_sc:.4f}
### Response:
"""
    return prompt
```

Generate Fonksiyonu

```python
def get_model_score(soru, dogru, ogrenci, sbert1, sbert2):
    p = make_prompt(soru, dogru, ogrenci, sbert1, sbert2)
    inputs = tokenizer_qwen([p], return_tensors="pt").to("cuda")

    outputs = model_qwen.generate(
    **inputs,
    max_new_tokens=250,
    use_cache=True,
    do_sample=False
    )
    full_text = tokenizer_qwen.batch_decode(outputs, skip_special_tokens=True)[0]

    return full_text.split("### Response:")[-1].strip() #sadece outputu dönderir
```

Verilen Stringten Puanı Bulan Fonksiyon

```python
def get_point(text):
    delimiter = "Puan:"

    if delimiter in text:
        parts = text.rsplit(delimiter, 1)
        comment = parts[0].strip()
        raw_score = parts[1].strip()

        try:
            # Puanı sayıya çevirmeyi deniyoruz
            score = float(raw_score)
        except ValueError:
            # Eğer sayıya çevrilemiyorsa (örn: "Belirsiz") metin olarak kalsın
            score = raw_score if raw_score else "Empty"
    else:
        comment = text.strip()
        score = None # "Bulunamadı" yerine None dönmek programatik olarak daha pratiktir

    return comment, score
```

**Tesseract**

```python
def tesseract_ocr(image_base64):
  img= decode_base64_image(image_base64)

  # Detaylı OCR (confidence skorlarıyla)
  ocr_data = pytesseract.image_to_data(
      img,
      lang='tur+eng',
      output_type=pytesseract.Output.DICT,
      config='--psm 3 --oem 3'  # PSM 3: Otomatik sayfa segmentasyonu
  )

  # Sadece metin
  text_simple = pytesseract.image_to_string(
      img,
      lang='tur+eng',
      config='--psm 3 --oem 3'
  )

  filtered_text = []
  detailed = []

  n_boxes = len(ocr_data['text'])
  for i in range(n_boxes):
      confidence = int(ocr_data['conf'][i])
      text = ocr_data['text'][i].strip()

      if confidence > 40 and text:
          filtered_text.append(text)
          detailed.append({
              "text": text,
              "confidence": confidence / 100.0,
              "bbox": [
                  ocr_data['left'][i],
                  ocr_data['top'][i],
                  ocr_data['width'][i],
                  ocr_data['height'][i]
              ]
          })

  full_text = ' '.join(filtered_text) if filtered_text else text_simple.strip()

  return full_text
```

```python
def decode_base64_image(b64_string):
    """Base64 string'i decode eder"""
    try:
        if 'base64,' in b64_string:
            b64_string = b64_string.split('base64,')[1]

        b64_string = re.sub(r'\s+', '', b64_string)

        missing_padding = len(b64_string) % 4
        if missing_padding:
            b64_string += '=' * (4 - missing_padding)

        img_bytes = base64.b64decode(b64_string)
        img = Image.open(io.BytesIO(img_bytes))

        if img.mode != 'RGB':
            img = img.convert('RGB')

        return img
    except Exception as e:
        raise ValueError(f"Base64 decode hatası: {str(e)}")
```

```python
@app.route('/send', methods=['POST']) #Ölü enpoint
def send():
    try:
        data = request.json
        if not data or "questions" not in data:
            return jsonify({"error": "Questions verisi bulunamadı"}), 400

        questions = data["questions"]
        all_texts = []

        # 1️⃣ Sıralamayı sabitle: Soru - Doğru Cevap - Öğrenci Cevabı
        for item in questions:
            all_texts.extend([
                item["question"],
                item["correct_answer"],
                item["student_answer"]
            ])

        # SBERT toplu işlem
        embeddings = modelSBERT.encode(all_texts, normalize_embeddings=True)

        results = []

        # 2️⃣ Verileri yeni sıralamaya göre çek
        for i in range(len(questions)):
            base_idx = i * 3

            q_emb = embeddings[base_idx]      # Soru
            dc_emb = embeddings[base_idx + 1] # Doğru Cevap
            s_emb = embeddings[base_idx + 2]  # Öğrenci Cevabı

            # Benzerlik Skorları (Skaler Çarpım)
            simQS = float(np.dot(q_emb, s_emb))  # Soru - Öğrenci
            simSC = float(np.dot(s_emb, dc_emb)) # Öğrenci - Doğru Cevap

            # Model Analizi (get_model_score: soru, dogru, ogrenci, s1, s2)
            model_output = get_model_score(
                questions[i]["question"],
                questions[i]["correct_answer"],
                questions[i]["student_answer"],
                simQS,
                simSC
            )

            results.append({
                "question_index": i + 1,
                "model_response": model_output
            })

            print(results[-1]["model_response"])

        return jsonify({
            "status": "success",
            "results": results
        }), 200

    except Exception as e:
        return jsonify({"error": str(e)}), 500
```

```python
@app.route('/ocr', methods=['POST'])
def ocr():
    try:
        data = request.json
        if not data or "question_sets" not in data:
            return jsonify({"error": "question_sets verisi bulunamadı"}), 400

        question_sets = data["question_sets"]
        if not isinstance(question_sets, list) or len(question_sets) == 0:
            return jsonify({"error": "question_sets boş veya geçersiz"}), 400

        results = []

        for i, qset in enumerate(question_sets):
            if "question" not in qset:
                return jsonify({"error": f"Set {i+1}: 'question' alanı eksik"}), 400
            if "correct_answer" not in qset:
                return jsonify({"error": f"Set {i+1}: 'correct_answer' alanı eksik"}), 400
            if "point" not in qset:
                return jsonify({"error" : f"Set {i+1}: 'point' alanı eksik"}), 400
            if "images" not in qset or len(qset["images"]) == 0:
                return jsonify({"error": f"Set {i+1}: 'images' alanı eksik veya boş"}), 400

            question       = qset["question"]
            correct_answer = qset["correct_answer"]
            point          = qset["point"]
            images         = qset["images"]

            # OCR: her fotoğraftan metin çıkar
            ocr_texts = []
            for image_base64 in images:
                ocr_texts.append(tesseract_ocr(image_base64))

            # SBERT: soru + doğru cevap + tüm öğrenci cevapları birlikte encode
            sbert_list = [question, correct_answer] + ocr_texts
            embeddings = modelSBERT.encode(sbert_list, normalize_embeddings=True)

            q_emb  = embeddings[0]   # soru
            dc_emb = embeddings[1]   # doğru cevap

            evaluation_list = []
            for k, ocr_text in enumerate(ocr_texts):
                s_emb = embeddings[k + 2]  # öğrenci cevabı

                simQS = float(np.dot(q_emb, s_emb))   # soru - öğrenci
                simSC = float(np.dot(s_emb, dc_emb))  # öğrenci - doğru cevap  ← simSC büyük harf

                model_output = get_model_score(
                    question,
                    correct_answer,
                    ocr_text,
                    simQS,
                    simSC
                )
                # ocr endpoint içindeki ilgili satırı şöyle güncelleyebilirsin:
                comment, ai_score = get_point(model_output)
                # Eğer ai_score sayı değilse 0.0 kabul et
                calculated_point = point * (ai_score if isinstance(ai_score, float) else 0.0)

                evaluation_list.append({
                    "image_index": k + 1,
                    "ocr_text": ocr_text,
                    "simQS": simQS,
                    "simSC": simSC,
                    "point": calculated_point,
                    "model_response": model_output
                })

            if not evaluation_list:
              evaluation_list = [{
                  "image_index": 0,
                  "ocr_text": "OCR çıktısı alınamadı",
                  "simQS": 0.0,
                  "simSC": 0.0,
                  "point": 0.0,
                  "model_response": "Değerlendirme yapılamadı"
              }]
            # ← döngü DIŞINDA, her soru için bir kez ekleniyor
            results.append({
                "question_index": i + 1,
                "question": question,
                "correct_answer": correct_answer,
                "analysis": evaluation_list
            })

        # ← for i döngüsü bittikten sonra döndür
        print(results)
        return jsonify({"results": results}), 200

    except Exception as e:
        return jsonify({"error": str(e)}), 500
```

```python
@app.route('/export-excel', methods=['POST'])
def export_excel():
    # Eski raporları temizle (10 dakikadan eski)
    for f in os.listdir('/tmp'):
        if f.startswith('rapor_') and f.endswith('.xlsx'):
            fpath = os.path.join('/tmp', f)
            if time.time() - os.path.getmtime(fpath) > 600:
                os.remove(fpath)

    data = request.json

    wb = Workbook()
    ws = wb.active
    ws.append(['Soru', 'Öğrenci', 'OCR Metni', 'Soru Uyumu %', 'Cevap Doğruluğu %', 'AI Yorumu', 'Puan'])

    for result in data['results']:
        question = result['question']
        for analysis in result['analysis']:
            raw_response = analysis['model_response']

            # --- Puanı Ayırma İşlemi ---
            comment = raw_response
            score = ""

            if "Puan:" in raw_response:
                comment, score = get_point(raw_response)
                # 'Puan:' kelimesine göre sağdan bir kez bölüyoruz
                #parts = raw_response.rsplit("Puan:", 1)
                #comment = parts[0].strip()  # Puan öncesindeki metin (AI Yorumu)
                #score = parts[1].strip()    # Puanın kendisi (x.xx)
            # ---------------------------

            ws.append([
                question,
                analysis['image_index'],
                analysis['ocr_text'],
                f"{analysis['simQS'] * 100:.1f}",
                f"{analysis['simSC'] * 100:.1f}",
                comment, # Temizlenmiş yorum
                analysis['point']    # Ayrılmış puan
            ])

    # Dosya kaydetme işlemleri...
    filename = f'rapor_{int(time.time())}.xlsx'
    filepath = os.path.join('/tmp', filename)
    wb.save(filepath)

    return jsonify({
        'download_url': f'{request.host_url}download/{filename}'
    })

# Download endpoint (aynı fiziksel dosyayı serve et)
@app.route('/download/<filename>')
def download_file(filename):
    return send_file(
        os.path.join('/tmp', filename),
        mimetype='application/vnd.openxmlformats-officedocument.spreadsheetml.sheet',
        download_name=filename,
        as_attachment=True
    )
```
